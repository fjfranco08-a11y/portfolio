**Least Squares Linear Regression**, that's a mouthful and quite confusing when not approached in pieces so let's break it down! 

_Linear regression_ ($X\vec{w}$ --> $y$) is a powerful _Hypothesis Function_ ($H(x)$ --> $y$) that we can use to model linear patterns in data. But before diving any deeper, it's vital we understand _Loss Functions_ + _Empirical Loss_ and how these all relate. 

_Loss Functions_ measure the difference between the $i$ th prediction $H(\vec{x}^{i})$ and the "right answer" $y_i$. A good hypothesis function makes good predictions and therefore minimizes average loss, aka _Empirical Loss_, for the entire dataset. 

The two types of loss functions I am familiar with are _Least Squares Loss_ and absolute loss. Using least squares loss and our linear regression hypothesis function, our loss function would be $\mathcal{l}_{sq} (H(\vec{x}^{i}), y_i) = (\vec{x}^{i}\vec{w} - y_i)^{2}$

With our loss function in hand, we can plug it into the Empirical risk definition, so 

$R(H) = \frac{1}{n} \sum_{i= 1}^{n} \mathcal{l} (H(\vec{x}^{i}), y_i)$ --> $R(H) = \frac{1}{n} \sum_{i= 1}^{n} (\vec{x}^{i}\vec{w} - y_i)^{2}$ 

Note, the hypothesis function is fully described by $\vec{w}$ since $X$ is our data matrix and remains constant so we can write $R(\vec{w})$ instead of $R(H)$. 

This is all nice and well but we soon encounter a problem. How do we know which linear function is best at minimizing our empirical risk? We can arbitrarily choose our weight vector $\vec{w}$ and get infinitely many $X\vec{w}$ to predict $y$. Now is the time to optimize! 



## 2. Optimization

Our goal is to minimize the empirical risk so that means deriving. By deriving we can find the gradient and where it is zero, aka where the maximum or minimum of the function is (this idea will be revisited later). 

As a reminder, our empirical risk formula is $R(\vec{w}) = \frac{1}{n} \sum_{i=1}^n(\vec{x}^i\vec{w} - y_{i})^{2}$. $X$ is our inputted data matrix and that remains fixed so the only thing to optimize is the weights vector $\vec{w}$ which means we are deriving with respect to $\vec{w}$. 

$\frac{\partial}{\partial \vec{w}} R(\vec{w})$ --> $\frac{\partial}{\partial \vec{w}}(\frac{1}{n} \sum_{i=1}^{n}(\vec{x}^i\vec{w} - y_i)^{2})$ --> $\frac{1}{n}\sum_{i=1}^{n} \frac{\partial}{\partial \vec{w}}(\vec{x}^{i}\vec{w} - y_i)^{2}$ 

Summing over all $i$ is equivalent to the matrix form $||\vec{y} - X\vec{w}||^{2}$ where each row of $X$ is $\vec{x}^i$, so we can simplify to $\frac{\partial}{\partial \vec{w}}||\vec{y} - X\vec{w}||^{2}$.


expand $||\vec{y} - X\vec{w}||^{2}$ into $||\vec{y}||^{2} - 2\vec{w}^{T}X^{T}\vec{y} + \vec{w}^{T}X{T}X\vec{w}$ 


$\frac{\partial}{\partial\vec{w}} (||y||^{2} - 2\vec{w}^{T}X^{T}\vec{y} + \vec{w}^{T}X{T}X\vec{w})$ --> $-2X^{T}\vec{y} + 2X^{T}X\vec{w}$

set $-2X^{T}\vec{y} + 2X^{T}X\vec{w}$ equal to 0 and isolate $\vec{w}$

$X^{T}\vec{y} = X^{T}X\vec{w}$ 

"divide" by $(X^{T}X)^{-1}$

$\vec{w} = (X^{T}X)^{-1}X^{T}\vec{y}$


## 3. Callback

Before implementing the code to test our solution, we should revist the portion that allowed us to reach such a conclusion, how did we know this was a minimum? 

Our function's minimum is when its second derivative is greater than 0 (since that means everywhere we slightly move to is going up). 

We know $\frac{\partial}{\partial \vec{w}} R(\vec{w})$ --> $-2X^{T}\vec{y} + 2X^{T}X\vec{w}$

so deriving once more with respect to $\vec{w}$ leaves us with $2X^{T}X$.

This is crucial since $X^{T}X$ is a positve semi-definite matrix which is proven to have a convex loss function. 



```python
import numpy as np

class LeastSquaresRegression:
    def __init__(self):
        self.w = None
    
    def fit(self, X, y):
        self.w = np.linalg.inv(X.T @ X) @ X.T @ y

    # a characteristic of the class (like an attribute) but usually has code executed like a method with the difference being METHODS HAVE PARAMETERS and PROPERTIES DO NOT
    @property
    def weights(self):
        return self.w
    
    def predict(self, X):
        return X @ self.w
```

```python
# Here we need to generate an input dataset using numpy to see if our custom LeastSquaresRegression works
bias_feature_vect = np.ones((100, 1))
# print(bias_feature_vect)
rand_feature_vect = np.random.rand(100, 1)
# print(rand_feature_vect)
X = np.hstack((bias_feature_vect, rand_feature_vect))
print(f'Generated X dataset sample: \n {X[:10]}')
```

```python
# Now time to generate the resulting output for each observation. Let's have our output (y) be correlated to the data matrix (X) using the function y = 2x + 3.
y = (X[:, 1] * 2) + 3
```

```python
# Finally, let's see if our LeastSquaresRegression class works intended by giving us the weights (3,2).
lin_model = LeastSquaresRegression()
lin_model.fit(X, y)
print(f'lin_model weights: \n {lin_model.weights}')
```
